```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# --- DONNÉES DE SECOURS (Si vous n'avez pas internet) ---
# Ce bloc génère un jeu de données fictif semblable à celui de SpaceX
np.random.seed(42)
n_flights = 90

data_simulee = {
    'FlightNumber': list(range(1, n_flights + 1)),
    'PayloadMass': np.random.uniform(1000, 15000, n_flights),
    'LaunchSite': np.random.choice(['CCAFS SLC 40', 'KSC LC 39A', 'VAFB SLC 4E'], n_flights),
    'Orbit': np.random.choice(['LEO', 'VLEO', 'GTO', 'PO', 'ISS'], n_flights),
    'Class': np.random.choice([0, 1], n_flights, p=[0.3, 0.7]), # 70% de réussite
    'Date': pd.date_range(start='2010-06-04', periods=n_flights, freq='ME')
}

df = pd.DataFrame(data_simulee)
# Extraction de l'année
df['Year'] = df['Date'].dt.year
# --------------------------------------------------------

# Configuration du style des graphiques
sns.set_theme(style="whitegrid")

# (Ensuite, collez la suite du code précédent pour afficher les graphiques...)

# Extraction de l'année pour le graphique linéaire
df['Year'] = pd.to_datetime(df['Date']).dt.year

# Configuration du style des graphiques
sns.set_theme(style="whitegrid")

# --- 1. SCATTER PLOT ---
# Relation entre Flight Number, Payload Mass et Class
plt.figure(figsize=(12, 6))
sns.scatterplot(data=df, x="FlightNumber", y="PayloadMass", hue="Class", palette="dark", alpha=0.8)
plt.title("Relation entre le numéro de vol et la masse de la charge utile selon le résultat")
plt.xlabel("Numéro de Vol (Flight Number)")
plt.ylabel("Masse de la charge utile (Payload Mass en kg)")
plt.legend(title="Succès (1=Oui, 0=Non)")
plt.show()

# --- 2. BAR CHARTS ---
# Comparaison du taux de succès par site de lancement
plt.figure(figsize=(10, 5))
df_site = df.groupby('LaunchSite')['Class'].mean().reset_index()
sns.barplot(data=df_site, x="LaunchSite", y="Class", palette="viridis")
plt.title("Taux de réussite par Site de Lancement")
plt.xlabel("Site de Lancement")
plt.ylabel("Taux de réussite")
plt.ylim(0, 1)
plt.show()

# Comparaison du taux de succès par type d'orbite
plt.figure(figsize=(12, 5))
df_orbit = df.groupby('Orbit')['Class'].mean().reset_index().sort_values(by="Class", ascending=False)
sns.barplot(data=df_orbit, x="Orbit", y="Class", palette="magma")
plt.title("Taux de réussite par Type d'Orbite")
plt.xlabel("Type d'Orbite")
plt.ylabel("Taux de réussite")
plt.ylim(0, 1)
plt.show()

# --- 3. LINE CHART ---
# Tendance annuelle du taux de réussite des atterrissages
plt.figure(figsize=(10, 5))
df_year = df.groupby('Year')['Class'].mean().reset_index()
sns.lineplot(data=df_year, x="Year", y="Class", marker="o", color="b", linewidth=2.5)
plt.title("Évolution annuelle du taux de réussite des atterrissages")
plt.xlabel("Année")
plt.ylabel("Taux de réussite moyen")
plt.ylim(0, 1)
plt.show()

# --- 4. CATPLOTS (SEABORN) ---
# Séparation des points par variables pour voir l'impact opérationnel
sns.catplot(data=df, x="FlightNumber", y="LaunchSite", hue="Class", aspect=2, kind="strip", palette="Set1")
plt.title("Impact du numéro de vol et du site de lancement sur la réussite")
plt.xlabel("Numéro de Vol")
plt.ylabel("Site de Lancement")
plt.show()
```
