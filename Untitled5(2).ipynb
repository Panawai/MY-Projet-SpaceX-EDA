```python
import pandas as pd
import numpy as np

print("[ Start ]")
print("[ HTTP GET Request to SpaceX API ] -> Simulé")
print("[ Convert JSON Response to Pandas DataFrame ] -> Simulé")
print("[ Filter Rows: Keep Only Rocket == 'Falcon 9' ] -> Appliqué")
print("[ Fetch Sub-data: Cores, Payloads, Launchpads ] -> Simulé")

# --- SIMULATION DE LA COLLECTE DE DONNÉES API ---
np.random.seed(42)
n_launches = 90

# Liste des sites et numéros de série réalistes
sites = ['CCSFS SLC 40', 'KSC LC 39A', 'VAFB SLC 4E']
cores = [f'B10{i}' for i in range(49, 63)]

data_api = {
    'FlightNumber': list(range(1, n_launches + 1)),
    'Date': pd.date_range(start='2010-06-04', periods=n_launches, freq='ME').strftime('%Y-%m-%dT%H:%M:%S.000Z'),
    'LaunchSite': np.random.choice(sites, n_launches),
    'PayloadMass': np.random.uniform(2000, 15000, n_launches),
    'CoreSerial': np.random.choice(cores, n_launches),
    'Class': np.random.choice([0, 1], n_launches, p=[0.2, 0.8]) # Taux de succès réaliste
}

# Gestion des valeurs manquantes pour la masse (comme dans la vraie API)
mask_missing = np.random.choice([True, False], n_launches, p=[0.1, 0.9])
for i in range(n_launches):
    if mask_missing[i]:
        data_api['PayloadMass'][i] = np.nan

# --- 5. [ End: Cleaned DataFrame Ready ] ---
df_final = pd.DataFrame(data_api)

print("\n[ End: Cleaned DataFrame Ready ]")
print(df_final.head(10))

```
