```python
import pandas as pd
import numpy as np

# --- 0. CRÉATION DU JEU DE DONNÉES BRUT DE DÉPART (Simulé pour Colab) ---
np.random.seed(42)
n_rows = 90

data_brute = {
    'FlightNumber': list(range(1, n_rows + 1)),
    'PayloadMass': np.random.uniform(2000, 15000, n_rows),
    'LandingOutcome': np.random.choice(['True RTLS', 'True ASDS', 'True Ocean', 'False ASDS', 'False RTLS', 'None None', 'None Ocean'], n_rows, p=[0.3, 0.4, 0.1, 0.1, 0.02, 0.06, 0.02]),
    'GridFins': np.random.choice([True, False], n_rows, p=[0.8, 0.2]),
    'Legs': np.random.choice([True, False], n_rows, p=[0.75, 0.25]),
    'Block': np.random.choice([1, 2, 3, 4, 5], n_rows, p=[0.1, 0.1, 0.2, 0.2, 0.4]),
    'ReusedCount': np.random.randint(0, 6, n_rows)
}

# Introduction volontaire de valeurs manquantes (Nulls) dans PayloadMass
mask_missing = np.random.choice([True, False], n_rows, p=[0.1, 0.9])
for i in range(n_rows):
    if mask_missing[i]:
        data_brute['PayloadMass'][i] = np.nan

df = pd.DataFrame(data_brute)

# --- 1. HANDLING MISSING VALUES (Traitement des valeurs manquantes) ---
# Calcul de la moyenne de PayloadMass pour remplacer les valeurs nulles
payload_mean = df['PayloadMass'].mean()
df['PayloadMass'] = df['PayloadMass'].fillna(payload_mean)

# --- 2 & 3. OUTCOME LABELING & BINARY ENCODING (Codage binaire de la cible) ---
# Définition de 1 pour les succès (RTLS, ASDS, Ocean) et 0 pour les échecs (None, False)
def determine_class(outcome):
    if 'True' in outcome:
        return 1
    else:
        return 0

df['Class'] = df['LandingOutcome'].apply(determine_class)

# --- 4. FEATURE SELECTION (Sélection des variables opérationnelles clés) ---
features_list = ['FlightNumber', 'PayloadMass', 'GridFins', 'Legs', 'Block', 'ReusedCount', 'Class']
df_wrangled = df[features_list].copy()

# --- AFFICHAGE DU RÉSULTAT NETTOYÉ ---
print("--- STATISTIQUES DES VALEURS MANQUANTES APRÈS TRAITEMENT ---")
print(df_wrangled.isnull().sum())
print("\n--- APERÇU DU DATAFRAME FINAL (DATA WRANGLING TERMINÉ) ---")
print(df_wrangled.head(10))
```
