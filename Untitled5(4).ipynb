```python
import sqlite3
import pandas as pd
import numpy as np

# --- 0. PRÉPARATION DE LA BASE DE DONNÉES EN MÉMOIRE (Simulé pour Colab) ---
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# Génération de données cohérentes pour exécuter les requêtes SQL
np.random.seed(42)
n_rows = 30
data_sql = {
    'FlightNumber': list(range(1, n_rows + 1)),
    'LaunchSite': np.random.choice(['CCAFS LC-40', 'CCAFS SLC-40', 'KSC LC-39A', 'VAFB SLC-4E'], n_rows),
    'Customer': np.random.choice(['NASA', 'SES', 'SpaceX', 'Iridium'], n_rows, p=[0.4, 0.2, 0.3, 0.1]),
    'PayloadMass': np.random.randint(2000, 15000, n_rows),
    'BoosterVersion': np.random.choice(['F9 v1.0', 'F9 v1.1', 'F9 B5'], n_rows),
    'CoreSerial': [f'B10{i}' for i in np.random.randint(49, 55, n_rows)],
    'Class': np.random.choice([1, 0], n_rows, p=[0.8, 0.2])
}
df_spacex = pd.DataFrame(data_sql)
df_spacex.to_sql('SPACEXTABLE', conn, index=False, if_exists='replace')

# --- EXÉCUTION DES REQUÊTES DE VOTRE DIAPOSITIVE ---

# 1. Data Selection: SELECT DISTINCT
print("--- 1. DATA SELECTION: SITES DE LANCEMENT UNIQUES ---")
query_1 = "SELECT DISTINCT LaunchSite FROM SPACEXTABLE;"
print(pd.read_sql_query(query_1, conn))
print("\n" + "="*50 + "\n")

# 2. Filtering and Aggregation: WHERE ... LIKE 'CCAFS%'
print("--- 2. FILTERING: ENREGISTREMENTS POUR LES SITES 'CCAFS' ---")
query_2 = "SELECT FlightNumber, LaunchSite, PayloadMass FROM SPACEXTABLE WHERE LaunchSite LIKE 'CCAFS%';"
print(pd.read_sql_query(query_2, conn).head())
print("\n" + "="*50 + "\n")

# 3. Calculations: SUM and AVG pour la NASA
print("--- 3. CALCULATIONS: MASSE TOTALE ET MOYENNE POUR LA NASA ---")
query_3 = """
SELECT SUM(PayloadMass) AS TotalPayloadMass, AVG(PayloadMass) AS AvgPayloadMass
FROM SPACEXTABLE
WHERE Customer = 'NASA';
"""
print(pd.read_sql_query(query_3, conn))
print("\n" + "="*50 + "\n")

# 4. Trend Analysis: GROUP BY (Succès et Échecs par version de Booster)
print("--- 4. TREND ANALYSIS: SUCCÈS / ÉCHECS PAR VERSION DE BOOSTER ---")
query_4 = """
SELECT BoosterVersion, Class, COUNT(Class) AS Count
FROM SPACEXTABLE
GROUP BY BoosterVersion, Class;
"""
print(pd.read_sql_query(query_4, conn))
print("\n" + "="*50 + "\n")

# 5. Subqueries: Numéros de série avec la masse maximale et atterrissage réussi (Class=1)
print("--- 5. SUBQUERIES: BOOSTERS AYANT RÉUSSI AVEC UNE CHARGE MAXIMALE ---")
query_5 = """
SELECT DISTINCT CoreSerial, PayloadMass
FROM SPACEXTABLE
WHERE Class = 1 AND PayloadMass = (SELECT MAX(PayloadMass) FROM SPACEXTABLE);
"""
print(pd.read_sql_query(query_5, conn))

# Fermeture de la connexion
conn.close()
```
