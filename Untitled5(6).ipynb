```python
import folium
from math import sin, cos, sqrt, atan2, radians

# --- FONCTION DE CALCUL DE DISTANCE (Haversine) ---
def calculate_distance(lat1, lon1, lat2, lon2):
    # Rayon de la Terre en kilomètres
    R = 6373.0

    lat1, lon1 = radians(lat1), radians(lon1)
    lat2, lon2 = radians(lat2), radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    return R * c

# --- 1. COORDINATE MARKERS & BASE MAP ---
# Initialisation de la carte centrée sur les États-Unis
nasa_map = folium.Map(location=[28.5623, -80.5774], zoom_start=5)

# Coordonnées réelles des sites de lancement SpaceX
launch_sites = {
    'CCAFS SLC-40': [28.562302, -80.577356],
    'KSC LC-39A': [28.573255, -80.646895],
    'VAFB SLC-4E': [34.632834, -120.610745]
}

# Coordonnées d'une infrastructure critique proche (Exemple : Côte Atlantique)
coastline_coord = [28.56367, -80.56801]

# Ajout des marqueurs et cercles de proximité
for site, coord in launch_sites.items():
    # Marqueur de texte personnalisé
    folium.Marker(
        location=coord,
        popup=site,
        icon=folium.Icon(color='blue', icon='info-sign')
    ).add_to(nasa_map)

    # --- 2. PROXIMITY CIRCLES ---
    # Cercles colorés pour évaluer visuellement la zone de sécurité
    folium.Circle(
        location=coord,
        radius=1000, # Rayon de 1 km
        color='#d35400',
        fill=True,
        fill_color='#d35400'
    ).add_to(nasa_map)

# --- 3. DISTANCE POLYLINES & SPATIAL ANALYSIS ---
# Calcul de la distance entre CCAFS SLC-40 et la côte
distance_coast = calculate_distance(launch_sites['CCAFS SLC-40'][0], launch_sites['CCAFS SLC-40'][1], coastline_coord[0], coastline_coord[1])

# Tracé de la ligne interactive entre le pas de tir et la côte
folium.PolyLine(
    locations=[launch_sites['CCAFS SLC-40'], coastline_coord],
    weight=2,
    color='blue'
).add_to(nasa_map)

# Ajout d'une étiquette textuelle indiquant la distance calculée
folium.Marker(
    location=coastline_coord,
    icon=folium.DivIcon(
        icon_size=(150,36),
        icon_anchor=(0,0),
        html=f'<div style="font-size: 12pt; color:#d35400;"><b>{distance_coast:.2f} KM</b></div>'
    )
).add_to(nasa_map)

# --- AFFICHAGE DE LA CARTE ---
# Note : Dans Google Colab, tapez simplement 'nasa_map' dans la cellule suivante pour l'afficher.
nasa_map
```
